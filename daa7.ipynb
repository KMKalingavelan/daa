```python
# Matrix Chain Multiplication
# Top-Down Memoization vs Bottom-Up DP
# Chain of 6 matrices


# --------------------------------------------------
# Matrix dimensions
# --------------------------------------------------

# 6 matrices:
# A1 = 10 x 20
# A2 = 20 x 30
# A3 = 30 x 40
# A4 = 40 x 30
# A5 = 30 x 20
# A6 = 20 x 10

dimensions = [10, 20, 30, 40, 30, 20, 10]

n = len(dimensions) - 1


# ==================================================
# 1. TOP-DOWN DP (MEMOIZATION)
# ==================================================

def matrix_chain_memo(dimensions):

    n = len(dimensions) - 1

    # Memoization table
    memo = [[None] * (n + 1) for _ in range(n + 1)]

    # Recursive function
    def solve(i, j):

        # Only one matrix
        if i == j:
            return 0

        # Return cached result
        if memo[i][j] is not None:
            return memo[i][j]

        # Initially infinity
        minimum_cost = float('inf')

        # Try every possible split
        for k in range(i, j):

            cost = (
                solve(i, k)
                + solve(k + 1, j)
                + dimensions[i - 1]
                * dimensions[k]
                * dimensions[j]
            )

            minimum_cost = min(minimum_cost, cost)

        # Store result in cache
        memo[i][j] = minimum_cost

        return minimum_cost

    minimum_cost = solve(1, n)

    return minimum_cost, memo


# ==================================================
# 2. BOTTOM-UP DP (TABULATION)
# ==================================================

def matrix_chain_bottom_up(dimensions):

    n = len(dimensions) - 1

    # DP table
    dp = [[0] * (n + 1) for _ in range(n + 1)]

    # Chain length
    for length in range(2, n + 1):

        for i in range(1, n - length + 2):

            j = i + length - 1

            dp[i][j] = float('inf')

            # Try every possible split
            for k in range(i, j):

                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + dimensions[i - 1]
                    * dimensions[k]
                    * dimensions[j]
                )

                dp[i][j] = min(dp[i][j], cost)

    return dp[1][n], dp


# ==================================================
# Run Top-Down Memoization
# ==================================================

top_down_cost, memo_table = matrix_chain_memo(dimensions)


# ==================================================
# Run Bottom-Up DP
# ==================================================

bottom_up_cost, dp_table = matrix_chain_bottom_up(dimensions)


# ==================================================
# Display Results
# ==================================================

print("Matrix Chain Multiplication")
print("===========================")

print("\nMatrix Chain:")

for i in range(n):
    print(
        f"A{i + 1} = "
        f"{dimensions[i]} x {dimensions[i + 1]}"
    )


print("\nTop-Down DP (Memoization)")
print("-------------------------")
print("Minimum Cost:", top_down_cost)


print("\nBottom-Up DP (Tabulation)")
print("-------------------------")
print("Minimum Cost:", bottom_up_cost)


# ==================================================
# Verify Both Results
# ==================================================

print("\nVerification")
print("------------")

if top_down_cost == bottom_up_cost:
    print("SUCCESS: Both approaches give the same minimum cost.")
else:
    print("ERROR: The results are different.")


# ==================================================
# Display Bottom-Up DP Table
# ==================================================

print("\nBottom-Up DP Table:")
print("-------------------")

for i in range(1, n + 1):
    print(dp_table[i][1:n + 1])
```

```python
# N-Queens Problem using Backtracking
# Find all solutions for N = 4
# Count total solutions for N = 8


# --------------------------------------------------
# Check whether a queen can be placed
# --------------------------------------------------

def is_safe(board, row, col, n):

    # Check previous rows
    for previous_row in range(row):

        previous_col = board[previous_row]

        # Same column
        if previous_col == col:
            return False

        # Same diagonal
        if abs(previous_row - row) == abs(previous_col - col):
            return False

    return True


# --------------------------------------------------
# Backtracking function
# --------------------------------------------------

def solve_n_queens(n):

    board = [-1] * n

    solutions = []

    def backtrack(row):

        # All queens have been placed
        if row == n:
            solutions.append(board.copy())
            return

        # Try every column in the current row
        for col in range(n):

            if is_safe(board, row, col, n):

                # Place queen
                board[row] = col

                # Move to next row
                backtrack(row + 1)

                # Remove queen (Backtrack)
                board[row] = -1

    backtrack(0)

    return solutions


# --------------------------------------------------
# Display a solution as a chessboard
# --------------------------------------------------

def display_solution(solution):

    n = len(solution)

    for row in range(n):

        for col in range(n):

            if solution[row] == col:
                print("Q", end=" ")
            else:
                print(".", end=" ")

        print()

    print()


# ==================================================
# PART 1: Find all solutions for N = 4
# ==================================================

n = 4

solutions_4 = solve_n_queens(n)

print("N-QUEENS PROBLEM")
print("================")

print("\nAll solutions for N = 4:")
print("------------------------")

for i, solution in enumerate(solutions_4, start=1):

    print(f"Solution {i}:")
    display_solution(solution)


print("Total solutions for N = 4:", len(solutions_4))


# ==================================================
# PART 2: Count solutions for N = 8
# ==================================================

n = 8

solutions_8 = solve_n_queens(n)

print("\nTotal solutions for N = 8:")
print("--------------------------")
print(len(solutions_8))
```

```python
# Knight's Tour using Backtracking
# 8 x 8 Chessboard
# Starting position: (0, 0)

N = 8

# Create an empty chessboard
board = [[-1 for _ in range(N)] for _ in range(N)]


# --------------------------------------------------
# All possible Knight moves
# --------------------------------------------------

move_x = [2, 1, -1, -2, -2, -1, 1, 2]
move_y = [1, 2, 2, 1, -1, -2, -2, -1]


# --------------------------------------------------
# Check whether a move is valid
# --------------------------------------------------

def is_safe(x, y):

    return (
        0 <= x < N and
        0 <= y < N and
        board[x][y] == -1
    )


# --------------------------------------------------
# Backtracking function
# --------------------------------------------------

def solve_knight_tour(x, y, move_count):

    # All 64 cells have been visited
    if move_count == N * N:
        return True

    # Try all possible Knight moves
    for i in range(8):

        next_x = x + move_x[i]
        next_y = y + move_y[i]

        if is_safe(next_x, next_y):

            # Mark the cell
            board[next_x][next_y] = move_count

            # Recursively continue
            if solve_knight_tour(
                next_x,
                next_y,
                move_count + 1
            ):
                return True

            # Backtrack
            board[next_x][next_y] = -1

    return False


# --------------------------------------------------
# Starting position
# --------------------------------------------------

start_x = 0
start_y = 0

# First cell is numbered 1
board[start_x][start_y] = 1


# --------------------------------------------------
# Solve the Knight's Tour
# --------------------------------------------------

if solve_knight_tour(start_x, start_y, 2):

    print("Knight's Tour Found!")
    print("====================")
    print("Starting Position: (0, 0)")
    print("\nNumbered Chessboard:\n")

    for row in board:
        print(" ".join(f"{cell:2}" for cell in row))

else:

    print("No Knight's Tour exists.")
```
