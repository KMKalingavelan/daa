```python
import heapq

# Number of vertices
V = 7

# List of edges (u, v, weight)
edges = [
    (0, 1, 7),
    (0, 3, 5),
    (1, 2, 8),
    (1, 3, 9),
    (1, 4, 7),
    (2, 4, 5),
    (3, 4, 15),
    (3, 5, 6),
    (4, 6, 9)
]

# ------------------ Kruskal's Algorithm ------------------

parent = list(range(V))
rank = [0] * V

def find(x):
    if parent[x] != x:
        parent[x] = find(parent[x])
    return parent[x]

def union(x, y):
    rootX = find(x)
    rootY = find(y)

    if rootX != rootY:
        if rank[rootX] < rank[rootY]:
            parent[rootX] = rootY
        elif rank[rootX] > rank[rootY]:
            parent[rootY] = rootX
        else:
            parent[rootY] = rootX
            rank[rootX] += 1

def kruskal():
    mst = []
    total_weight = 0

    sorted_edges = sorted(edges, key=lambda x: x[2])

    for u, v, w in sorted_edges:
        if find(u) != find(v):
            union(u, v)
            mst.append((u, v, w))
            total_weight += w

    return mst, total_weight

# ------------------ Prim's Algorithm ------------------

def prim():
    graph = [[] for _ in range(V)]

    for u, v, w in edges:
        graph[u].append((v, w))
        graph[v].append((u, w))

    visited = [False] * V
    pq = [(0, 0, -1)]  # (weight, current_vertex, parent)

    mst = []
    total_weight = 0

    while pq:
        weight, u, parent = heapq.heappop(pq)

        if visited[u]:
            continue

        visited[u] = True

        if parent != -1:
            mst.append((parent, u, weight))
            total_weight += weight

        for neighbor, w in graph[u]:
            if not visited[neighbor]:
                heapq.heappush(pq, (w, neighbor, u))

    return mst, total_weight

# ------------------ Main ------------------

kruskal_mst, kruskal_weight = kruskal()
prim_mst, prim_weight = prim()

print("Kruskal's MST:")
for u, v, w in kruskal_mst:
    print(f"{u} -- {v} == {w}")
print("Total Weight:", kruskal_weight)

print("\nPrim's MST:")
for u, v, w in prim_mst:
    print(f"{u} -- {v} == {w}")
print("Total Weight:", prim_weight)

print("\nVerification:")
if kruskal_weight == prim_weight:
    print("Both algorithms produce the SAME Minimum Spanning Tree weight.")
else:
    print("The MST weights are DIFFERENT.")

print("\nTime Complexity:")
print("Kruskal's Algorithm : O(E log E)")
print("Prim's Algorithm    : O(E log V) using Priority Queue")
```
