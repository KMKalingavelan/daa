```python
# ==========================================================
# BIN PACKING APPROXIMATION ALGORITHMS
#
# Algorithms:
# 1. Next Fit
# 2. First Fit
# 3. First Fit Decreasing
#
# Capacity = 1.0
# ==========================================================

import math


# ----------------------------------------------------------
# Input
# ----------------------------------------------------------

weights = [
    0.5, 0.7, 0.3, 0.9, 0.2,
    0.6, 0.8, 0.4, 0.1, 0.5
]

capacity = 1.0


# ==========================================================
# 1. THEORETICAL LOWER BOUND
# ==========================================================

total_weight = sum(weights)

lower_bound = math.ceil(
    total_weight / capacity
)

print("BIN PACKING PROBLEM")
print("===================")

print("\nItems:")
print(weights)

print("\nBin Capacity:", capacity)

print("Total Weight:", total_weight)

print("Theoretical Lower Bound:",
      lower_bound, "bins")


# ==========================================================
# 2. NEXT FIT
# ==========================================================

def next_fit(items, capacity):

    bins = []

    current_bin = []

    current_weight = 0

    for item in items:

        # If item fits in current bin
        if current_weight + item <= capacity:

            current_bin.append(item)

            current_weight += item

        else:

            # Close current bin
            bins.append(current_bin)

            # Start new bin
            current_bin = [item]

            current_weight = item

    # Add final bin
    if current_bin:
        bins.append(current_bin)

    return bins


# ==========================================================
# 3. FIRST FIT
# ==========================================================

def first_fit(items, capacity):

    bins = []

    bin_weights = []

    for item in items:

        placed = False

        # Try every existing bin
        for i in range(len(bins)):

            if bin_weights[i] + item <= capacity:

                bins[i].append(item)

                bin_weights[i] += item

                placed = True

                break

        # If item did not fit anywhere
        if not placed:

            bins.append([item])

            bin_weights.append(item)

    return bins


# ==========================================================
# 4. FIRST FIT DECREASING
# ==========================================================

def first_fit_decreasing(items, capacity):

    # Sort items in decreasing order
    sorted_items = sorted(
        items,
        reverse=True
    )

    return first_fit(
        sorted_items,
        capacity
    )


# ==========================================================
# 5. RUN ALGORITHMS
# ==========================================================

nf_bins = next_fit(
    weights,
    capacity
)

ff_bins = first_fit(
    weights,
    capacity
)

ffd_bins = first_fit_decreasing(
    weights,
    capacity
)


# ==========================================================
# 6. DISPLAY BIN CONTENTS
# ==========================================================

def display_bins(name, bins):

    print("\n" + name)
    print("-" * len(name))

    for i, bin_items in enumerate(bins, start=1):

        print(
            f"Bin {i}: {bin_items}"
            f"  -> Weight = {sum(bin_items):.1f}"
        )

    print("Total Bins:", len(bins))

    difference = len(bins) - lower_bound

    print(
        "Above Lower Bound:",
        difference,
        "bin(s)"
    )


# ==========================================================
# 7. DISPLAY RESULTS
# ==========================================================

display_bins(
    "1. NEXT FIT",
    nf_bins
)

display_bins(
    "2. FIRST FIT",
    ff_bins
)

display_bins(
    "3. FIRST FIT DECREASING",
    ffd_bins
)


# ==========================================================
# 8. FINAL COMPARISON
# ==========================================================

print("\n\nFINAL COMPARISON")
print("================")

print(
    f"{'Algorithm':<25}"
    f"{'Bins Used':<15}"
    f"{'Lower Bound':<15}"
    f"{'Extra Bins'}"
)

print("-" * 70)

print(
    f"{'Next Fit':<25}"
    f"{len(nf_bins):<15}"
    f"{lower_bound:<15}"
    f"{len(nf_bins) - lower_bound}"
)

print(
    f"{'First Fit':<25}"
    f"{len(ff_bins):<15}"
    f"{lower_bound:<15}"
    f"{len(ff_bins) - lower_bound}"
)

print(
    f"{'First Fit Decreasing':<25}"
    f"{len(ffd_bins):<15}"
    f"{lower_bound:<15}"
    f"{len(ffd_bins) - lower_bound}"
)
```

```python
# ==========================================================
# CLOUD SERVER RESOURCE ALLOCATION
# FFD vs BFD
#
# Server Capacity = 8 GB RAM
# ==========================================================

# Task RAM requirements
tasks = [
    1.5, 2.0, 3.5, 0.5, 4.0,
    1.0, 2.5, 3.0, 1.5, 2.0,
    0.5, 4.0, 1.0, 3.5, 2.0
]

capacity = 8.0


# ==========================================================
# 1. FIRST FIT DECREASING (FFD)
# ==========================================================

def first_fit_decreasing(tasks, capacity):

    # Sort tasks from largest to smallest
    sorted_tasks = sorted(
        tasks,
        reverse=True
    )

    servers = []
    server_usage = []

    for task in sorted_tasks:

        placed = False

        # Try existing servers
        for i in range(len(servers)):

            if server_usage[i] + task <= capacity:

                servers[i].append(task)

                server_usage[i] += task

                placed = True

                break

        # Create new server if task cannot fit
        if not placed:

            servers.append([task])

            server_usage.append(task)

    return servers, server_usage


# ==========================================================
# 2. BEST FIT DECREASING (BFD)
# ==========================================================

def best_fit_decreasing(tasks, capacity):

    # Sort tasks from largest to smallest
    sorted_tasks = sorted(
        tasks,
        reverse=True
    )

    servers = []
    server_usage = []

    for task in sorted_tasks:

        best_server = -1

        best_remaining = float("inf")

        # Find the server with the smallest
        # remaining space after placing the task
        for i in range(len(servers)):

            remaining = (
                capacity
                - server_usage[i]
                - task
            )

            if remaining >= 0 and remaining < best_remaining:

                best_remaining = remaining

                best_server = i

        # Place task in best server
        if best_server != -1:

            servers[best_server].append(task)

            server_usage[best_server] += task

        else:

            # Create a new server
            servers.append([task])

            server_usage.append(task)

    return servers, server_usage


# ==========================================================
# 3. RUN FFD
# ==========================================================

ffd_servers, ffd_usage = first_fit_decreasing(
    tasks,
    capacity
)


# ==========================================================
# 4. RUN BFD
# ==========================================================

bfd_servers, bfd_usage = best_fit_decreasing(
    tasks,
    capacity
)


# ==========================================================
# 5. DISPLAY FUNCTION
# ==========================================================

def display_result(name, servers, usage):

    print("\n" + name)
    print("=" * len(name))

    for i in range(len(servers)):

        print(
            f"Server {i + 1}: "
            f"{servers[i]}"
        )

        print(
            f"  RAM Used: {usage[i]:.1f} GB"
            f" / {capacity:.1f} GB"
        )

        print(
            f"  RAM Free: "
            f"{capacity - usage[i]:.1f} GB"
        )

    print(
        "\nTotal Servers Required:",
        len(servers)
    )


# ==========================================================
# 6. DISPLAY RESULTS
# ==========================================================

print("CLOUD SERVER RESOURCE ALLOCATION")
print("=================================")

print("\nServer Capacity:", capacity, "GB")

print("\nTask RAM Requirements:")
print(tasks)

print("\nTotal RAM Required:",
      sum(tasks), "GB")


display_result(
    "FIRST FIT DECREASING (FFD)",
    ffd_servers,
    ffd_usage
)


display_result(
    "BEST FIT DECREASING (BFD)",
    bfd_servers,
    bfd_usage
)


# ==========================================================
# 7. FINAL COMPARISON
# ==========================================================

print("\n\nFINAL COMPARISON")
print("================")

print(
    "FFD Servers:",
    len(ffd_servers)
)

print(
    "BFD Servers:",
    len(bfd_servers)
)

if len(ffd_servers) == len(bfd_servers):

    print(
        "\nBoth algorithms require the same number "
        "of servers."
    )

elif len(ffd_servers) < len(bfd_servers):

    print(
        "\nFFD uses fewer servers."
    )

else:

    print(
        "\nBFD uses fewer servers."
    )
```

```python
# ==========================================================
# BIN PACKING:
# NFD vs FFD vs BFD
#
# 50 random items
# Item sizes: 0.1 to 0.9
# Bin capacity: 1.0
# Number of trials: 100
# ==========================================================

import random


# ----------------------------------------------------------
# Parameters
# ----------------------------------------------------------

BIN_CAPACITY = 1.0
NUM_ITEMS = 50
NUM_TRIALS = 100

# Fix seed so results are reproducible
random.seed(42)


# ==========================================================
# 1. NEXT FIT DECREASING (NFD)
# ==========================================================

def next_fit_decreasing(items, capacity):

    # Sort items from largest to smallest
    sorted_items = sorted(items, reverse=True)

    bins = []

    current_bin_weight = 0.0

    for item in sorted_items:

        # If item fits in the current bin
        if current_bin_weight + item <= capacity:

            current_bin_weight += item

        else:

            # Close current bin and start a new one
            bins.append(current_bin_weight)

            current_bin_weight = item

    # Add the last bin
    if current_bin_weight > 0:
        bins.append(current_bin_weight)

    return len(bins)


# ==========================================================
# 2. FIRST FIT DECREASING (FFD)
# ==========================================================

def first_fit_decreasing(items, capacity):

    # Sort from largest to smallest
    sorted_items = sorted(items, reverse=True)

    bins = []

    for item in sorted_items:

        placed = False

        # Try existing bins in order
        for i in range(len(bins)):

            if bins[i] + item <= capacity:

                bins[i] += item

                placed = True

                break

        # Create new bin if necessary
        if not placed:

            bins.append(item)

    return len(bins)


# ==========================================================
# 3. BEST FIT DECREASING (BFD)
# ==========================================================

def best_fit_decreasing(items, capacity):

    # Sort from largest to smallest
    sorted_items = sorted(items, reverse=True)

    bins = []

    for item in sorted_items:

        best_bin = -1
        smallest_remaining = float("inf")

        # Find the bin that leaves the least
        # remaining space after placing the item
        for i in range(len(bins)):

            remaining = capacity - (bins[i] + item)

            if remaining >= -1e-9:

                if remaining < smallest_remaining:

                    smallest_remaining = remaining
                    best_bin = i

        # Place item in best bin
        if best_bin != -1:

            bins[best_bin] += item

        else:

            # Create a new bin
            bins.append(item)

    return len(bins)


# ==========================================================
# 4. RUN 100 RANDOM TRIALS
# ==========================================================

nfd_total = 0
ffd_total = 0
bfd_total = 0

nfd_results = []
ffd_results = []
bfd_results = []


for trial in range(NUM_TRIALS):

    # Generate 50 random items between 0.1 and 0.9
    items = [
        random.uniform(0.1, 0.9)
        for _ in range(NUM_ITEMS)
    ]

    # Apply the three algorithms
    nfd_bins = next_fit_decreasing(
        items,
        BIN_CAPACITY
    )

    ffd_bins = first_fit_decreasing(
        items,
        BIN_CAPACITY
    )

    bfd_bins = best_fit_decreasing(
        items,
        BIN_CAPACITY
    )

    # Store results
    nfd_results.append(nfd_bins)
    ffd_results.append(ffd_bins)
    bfd_results.append(bfd_bins)

    nfd_total += nfd_bins
    ffd_total += ffd_bins
    bfd_total += bfd_bins


# ==========================================================
# 5. CALCULATE AVERAGES
# ==========================================================

nfd_average = nfd_total / NUM_TRIALS
ffd_average = ffd_total / NUM_TRIALS
bfd_average = bfd_total / NUM_TRIALS


# ==========================================================
# 6. DISPLAY RESULTS
# ==========================================================

print("BIN PACKING APPROXIMATION COMPARISON")
print("====================================")

print("\nNumber of items per trial :", NUM_ITEMS)
print("Number of trials          :", NUM_TRIALS)
print("Bin capacity              :", BIN_CAPACITY)
print("Item range                : 0.1 to 0.9")


print("\n\nAVERAGE NUMBER OF BINS")
print("----------------------")

print(
    f"NFD (Next Fit Decreasing)  : "
    f"{nfd_average:.2f}"
)

print(
    f"FFD (First Fit Decreasing) : "
    f"{ffd_average:.2f}"
)

print(
    f"BFD (Best Fit Decreasing)  : "
    f"{bfd_average:.2f}"
)


# ==========================================================
# 7. COMPARISON
# ==========================================================

print("\n\nCOMPARISON")
print("----------")

print(
    "NFD vs FFD improvement:",
    f"{nfd_average - ffd_average:.2f} bins"
)

print(
    "NFD vs BFD improvement:",
    f"{nfd_average - bfd_average:.2f} bins"
)

print(
    "FFD vs BFD difference:",
    f"{ffd_average - bfd_average:.2f} bins"
)


# Find best algorithm
averages = {
    "NFD": nfd_average,
    "FFD": ffd_average,
    "BFD": bfd_average
}

best_algorithm = min(
    averages,
    key=averages.get
)

print(
    "\nBest average performance:",
    best_algorithm
)
```
