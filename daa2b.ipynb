```python
# KMP Algorithm Implementation

def compute_lps(pattern):
    """
    Compute LPS (Longest Prefix Suffix) array
    """
    lps = [0] * len(pattern)

    length = 0
    i = 1

    while i < len(pattern):
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1

        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1

    return lps


def kmp_search(text, pattern):
    lps = compute_lps(pattern)

    print("Pattern:", pattern)
    print("LPS Array:", lps)
    print()

    i = 0  # index for text
    j = 0  # index for pattern

    shifts = []

    while i < len(text):

        print(f"Comparing Text[{i}]='{text[i]}' with Pattern[{j}]='{pattern[j]}'")

        if text[i] == pattern[j]:
            i += 1
            j += 1

            if j == len(pattern):
                position = i - j
                print("\nPattern found at index:", position)
                shifts.append(position)

                j = lps[j - 1]

        else:
            if j != 0:
                old_j = j
                j = lps[j - 1]
                print(f"Mismatch: Shift pattern using LPS from {old_j} to {j}")
            else:
                i += 1
                print("Mismatch: Move to next text character")

        print()

    if not shifts:
        print("Pattern not found")

    return shifts


# Given input
text = "AABCAABXAAAZ"
pattern = "AABXAA"

# Run KMP Algorithm
kmp_search(text, pattern)
```
