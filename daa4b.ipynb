```python
# Dijkstra's Algorithm
# 5-Node City Road Network
# Node 0 = Airport
# Node 4 = Central Station

import heapq

def dijkstra(graph, start, destination):
    # Number of nodes
    n = len(graph)

    # Initially, distance to every node is infinity
    distance = [float('inf')] * n

    # Store the previous node for finding the shortest path
    parent = [-1] * n

    # Distance from source to itself is 0
    distance[start] = 0

    # Priority queue: (distance, node)
    priority_queue = [(0, start)]

    while priority_queue:

        current_distance, current_node = heapq.heappop(priority_queue)

        # Ignore outdated entries
        if current_distance > distance[current_node]:
            continue

        # Explore all outgoing roads
        for neighbor, travel_time in graph[current_node]:

            new_distance = current_distance + travel_time

            # Relaxation
            if new_distance < distance[neighbor]:
                distance[neighbor] = new_distance
                parent[neighbor] = current_node

                heapq.heappush(
                    priority_queue,
                    (new_distance, neighbor)
                )

    # Reconstruct the shortest path
    path = []
    current = destination

    while current != -1:
        path.append(current)
        current = parent[current]

    path.reverse()

    return distance[destination], path


# --------------------------------------------------
# City Road Network
# --------------------------------------------------

# graph[node] = [(neighbor, travel_time), ...]

graph = {
    0: [(1, 4), (2, 2)],
    1: [(2, 1), (3, 5)],
    2: [(1, 1), (3, 8), (4, 10)],
    3: [(4, 2)],
    4: []
}


# --------------------------------------------------
# Find fastest route
# --------------------------------------------------

source = 0       # Airport
destination = 4  # Central Station

shortest_time, shortest_path = dijkstra(
    graph,
    source,
    destination
)


# --------------------------------------------------
# Display Result
# --------------------------------------------------

print("Dijkstra's Algorithm")
print("--------------------")

print("Source: Airport (Node 0)")
print("Destination: Central Station (Node 4)")

print("\nFastest Route:")

for i, node in enumerate(shortest_path):
    if i < len(shortest_path) - 1:
        print(node, end=" -> ")
    else:
        print(node)

print("\nTotal Travel Time:", shortest_time, "minutes")
```

```python
# Dijkstra's Algorithm with Hop Count
# If two paths have the same total weight,
# choose the path with fewer edges (hops).

import heapq


def dijkstra_with_hops(graph, source, destination):

    n = len(graph)

    # distance[i] = shortest distance from source to i
    distance = [float('inf')] * n

    # hops[i] = number of edges in the selected shortest path
    hops = [float('inf')] * n

    # parent[i] = previous node in the shortest path
    parent = [-1] * n

    # Starting node
    distance[source] = 0
    hops[source] = 0

    # Priority queue stores:
    # (distance, hops, node)
    priority_queue = [(0, 0, source)]

    while priority_queue:

        current_distance, current_hops, current_node = heapq.heappop(
            priority_queue
        )

        # Ignore outdated entries
        if (current_distance > distance[current_node] or
            (current_distance == distance[current_node] and
             current_hops > hops[current_node])):
            continue

        # Explore all adjacent nodes
        for neighbor, weight in graph[current_node]:

            new_distance = current_distance + weight
            new_hops = current_hops + 1

            # Update if:
            # 1. New path has smaller total weight
            # OR
            # 2. Same weight but fewer hops
            if (new_distance < distance[neighbor] or
                (new_distance == distance[neighbor] and
                 new_hops < hops[neighbor])):

                distance[neighbor] = new_distance
                hops[neighbor] = new_hops
                parent[neighbor] = current_node

                heapq.heappush(
                    priority_queue,
                    (new_distance, new_hops, neighbor)
                )

    # Reconstruct shortest path
    path = []
    current = destination

    while current != -1:
        path.append(current)
        current = parent[current]

    path.reverse()

    return distance[destination], hops[destination], path


# --------------------------------------------------
# 6-Vertex Directed Weighted Graph
# --------------------------------------------------

graph = {
    0: [(1, 2), (2, 5)],
    1: [(2, 3), (3, 4)],
    2: [(3, 1), (4, 5)],
    3: [(4, 2), (5, 4)],
    4: [(5, 2)],
    5: []
}


# --------------------------------------------------
# Find shortest path from node 0 to node 5
# --------------------------------------------------

source = 0
destination = 5

shortest_distance, shortest_hops, shortest_path = dijkstra_with_hops(
    graph,
    source,
    destination
)


# --------------------------------------------------
# Display Result
# --------------------------------------------------

print("Dijkstra's Algorithm with Hop Count")
print("------------------------------------")

print("Source      :", source)
print("Destination :", destination)

print("\nShortest Path:")

for i, node in enumerate(shortest_path):
    if i < len(shortest_path) - 1:
        print(node, end=" -> ")
    else:
        print(node)

print("\nTotal Weight :", shortest_distance)
print("Hop Count    :", shortest_hops)
```
