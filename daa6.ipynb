```python
# Matrix Chain Multiplication using Dynamic Programming
# Matrices:
# A1 = 10 x 30
# A2 = 30 x 5
# A3 = 5 x 60
# A4 = 60 x 10


# --------------------------------------------------
# Matrix dimensions
# --------------------------------------------------

# A1 = 10 x 30
# A2 = 30 x 5
# A3 = 5 x 60
# A4 = 60 x 10

dimensions = [10, 30, 5, 60, 10]

n = len(dimensions) - 1


# --------------------------------------------------
# DP tables
# --------------------------------------------------

# m[i][j] = minimum number of scalar multiplications
# required to multiply Ai...Aj

m = [[0] * (n + 1) for _ in range(n + 1)]

# s[i][j] = position where the optimal split occurs
s = [[0] * (n + 1) for _ in range(n + 1)]


# --------------------------------------------------
# Matrix Chain Multiplication
# --------------------------------------------------

for length in range(2, n + 1):

    for i in range(1, n - length + 2):

        j = i + length - 1

        m[i][j] = float('inf')

        for k in range(i, j):

            cost = (
                m[i][k]
                + m[k + 1][j]
                + dimensions[i - 1]
                * dimensions[k]
                * dimensions[j]
            )

            if cost < m[i][j]:

                m[i][j] = cost
                s[i][j] = k


# --------------------------------------------------
# Construct optimal multiplication order
# --------------------------------------------------

def print_optimal_order(s, i, j):

    if i == j:
        return f"A{i}"

    k = s[i][j]

    left = print_optimal_order(s, i, k)
    right = print_optimal_order(s, k + 1, j)

    return f"({left} × {right})"


optimal_order = print_optimal_order(s, 1, n)


# --------------------------------------------------
# Display results
# --------------------------------------------------

print("Matrix Chain Multiplication")
print("---------------------------")

print("\nMatrix Dimensions:")
print("A1 = 10 x 30")
print("A2 = 30 x 5")
print("A3 = 5 x 60")
print("A4 = 60 x 10")

print("\nOptimal Multiplication Order:")
print(optimal_order)

print("\nMinimum Number of Scalar Multiplications:")
print(m[1][n])


# --------------------------------------------------
# Display DP Table
# --------------------------------------------------

print("\nDP Table:")
print("--------")

for i in range(1, n + 1):
    print(m[i][1:n + 1])
```

```python
# Matrix Chain Multiplication using Dynamic Programming
#
# Matrices:
# T = 4 x 4
# R = 4 x 4
# S = 4 x 4
# V = 4 x 10000
# P = 10000 x 1


# --------------------------------------------------
# Matrix dimensions
# --------------------------------------------------

# T(4x4), R(4x4), S(4x4), V(4x10000), P(10000x1)

dimensions = [4, 4, 4, 4, 10000, 1]

n = len(dimensions) - 1


# --------------------------------------------------
# DP Tables
# --------------------------------------------------

# m[i][j] = minimum scalar multiplications
# required to multiply matrices i through j

m = [[0] * (n + 1) for _ in range(n + 1)]

# s[i][j] = optimal splitting position
s = [[0] * (n + 1) for _ in range(n + 1)]


# --------------------------------------------------
# Matrix Chain Multiplication DP
# --------------------------------------------------

for length in range(2, n + 1):

    for i in range(1, n - length + 2):

        j = i + length - 1

        m[i][j] = float('inf')

        for k in range(i, j):

            cost = (
                m[i][k]
                + m[k + 1][j]
                + dimensions[i - 1]
                * dimensions[k]
                * dimensions[j]
            )

            if cost < m[i][j]:

                m[i][j] = cost
                s[i][j] = k


# --------------------------------------------------
# Construct Optimal Order
# --------------------------------------------------

def optimal_order(s, i, j):

    if i == j:
        matrix_names = ["", "T", "R", "S", "V", "P"]
        return matrix_names[i]

    k = s[i][j]

    left = optimal_order(s, i, k)
    right = optimal_order(s, k + 1, j)

    return "(" + left + " × " + right + ")"


# --------------------------------------------------
# Display Results
# --------------------------------------------------

print("Matrix Chain Multiplication using Dynamic Programming")
print("------------------------------------------------------")

print("\nMatrices:")
print("T = 4 x 4")
print("R = 4 x 4")
print("S = 4 x 4")
print("V = 4 x 10000")
print("P = 10000 x 1")

print("\nOptimal Multiplication Order:")
print(optimal_order(s, 1, n))

print("\nMinimum Number of Scalar Multiplications:")
print(m[1][n])


# --------------------------------------------------
# Display DP Table
# --------------------------------------------------

print("\nDP Table:")
print("--------")

for i in range(1, n + 1):
    print(m[i][1:n + 1])
```

```python
# Matrix Chain Multiplication
# Top-Down Memoization vs Bottom-Up DP
# Chain of 6 matrices


# --------------------------------------------------
# Matrix dimensions
# --------------------------------------------------

# 6 matrices:
# A1 = 10 x 20
# A2 = 20 x 30
# A3 = 30 x 40
# A4 = 40 x 30
# A5 = 30 x 20
# A6 = 20 x 10

dimensions = [10, 20, 30, 40, 30, 20, 10]

n = len(dimensions) - 1


# ==================================================
# 1. TOP-DOWN DP (MEMOIZATION)
# ==================================================

def matrix_chain_memo(dimensions):

    n = len(dimensions) - 1

    # Memoization table
    memo = [[None] * (n + 1) for _ in range(n + 1)]

    # Recursive function
    def solve(i, j):

        # Only one matrix
        if i == j:
            return 0

        # Return cached result
        if memo[i][j] is not None:
            return memo[i][j]

        # Initially infinity
        minimum_cost = float('inf')

        # Try every possible split
        for k in range(i, j):

            cost = (
                solve(i, k)
                + solve(k + 1, j)
                + dimensions[i - 1]
                * dimensions[k]
                * dimensions[j]
            )

            minimum_cost = min(minimum_cost, cost)

        # Store result in cache
        memo[i][j] = minimum_cost

        return minimum_cost

    minimum_cost = solve(1, n)

    return minimum_cost, memo


# ==================================================
# 2. BOTTOM-UP DP (TABULATION)
# ==================================================

def matrix_chain_bottom_up(dimensions):

    n = len(dimensions) - 1

    # DP table
    dp = [[0] * (n + 1) for _ in range(n + 1)]

    # Chain length
    for length in range(2, n + 1):

        for i in range(1, n - length + 2):

            j = i + length - 1

            dp[i][j] = float('inf')

            # Try every possible split
            for k in range(i, j):

                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + dimensions[i - 1]
                    * dimensions[k]
                    * dimensions[j]
                )

                dp[i][j] = min(dp[i][j], cost)

    return dp[1][n], dp


# ==================================================
# Run Top-Down Memoization
# ==================================================

top_down_cost, memo_table = matrix_chain_memo(dimensions)


# ==================================================
# Run Bottom-Up DP
# ==================================================

bottom_up_cost, dp_table = matrix_chain_bottom_up(dimensions)


# ==================================================
# Display Results
# ==================================================

print("Matrix Chain Multiplication")
print("===========================")

print("\nMatrix Chain:")

for i in range(n):
    print(
        f"A{i + 1} = "
        f"{dimensions[i]} x {dimensions[i + 1]}"
    )


print("\nTop-Down DP (Memoization)")
print("-------------------------")
print("Minimum Cost:", top_down_cost)


print("\nBottom-Up DP (Tabulation)")
print("-------------------------")
print("Minimum Cost:", bottom_up_cost)


# ==================================================
# Verify Both Results
# ==================================================

print("\nVerification")
print("------------")

if top_down_cost == bottom_up_cost:
    print("SUCCESS: Both approaches give the same minimum cost.")
else:
    print("ERROR: The results are different.")


# ==================================================
# Display Bottom-Up DP Table
# ==================================================

print("\nBottom-Up DP Table:")
print("-------------------")

for i in range(1, n + 1):
    print(dp_table[i][1:n + 1])
```
