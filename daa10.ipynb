```python
# ==========================================================
# DETERMINISTIC vs RANDOMIZED QUICK SORT
# Array Size = 10,000
# ==========================================================

import random
import time


# ----------------------------------------------------------
# Global comparison counters
# ----------------------------------------------------------

det_comparisons = 0
rand_comparisons = 0


# ==========================================================
# 1. DETERMINISTIC QUICK SORT
#    Pivot = Last Element
# ==========================================================

def deterministic_partition(arr, low, high):

    global det_comparisons

    pivot = arr[high]

    i = low - 1

    for j in range(low, high):

        det_comparisons += 1

        if arr[j] <= pivot:

            i += 1

            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


def deterministic_quick_sort(arr, low, high):

    if low < high:

        pivot_index = deterministic_partition(
            arr, low, high
        )

        deterministic_quick_sort(
            arr, low, pivot_index - 1
        )

        deterministic_quick_sort(
            arr, pivot_index + 1, high
        )


# ==========================================================
# 2. RANDOMIZED QUICK SORT
#    Pivot = Random Element
# ==========================================================

def randomized_partition(arr, low, high):

    global rand_comparisons

    # Select random pivot
    random_index = random.randint(low, high)

    # Move random pivot to the last position
    arr[random_index], arr[high] = (
        arr[high],
        arr[random_index]
    )

    pivot = arr[high]

    i = low - 1

    for j in range(low, high):

        rand_comparisons += 1

        if arr[j] <= pivot:

            i += 1

            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = (
        arr[high],
        arr[i + 1]
    )

    return i + 1


def randomized_quick_sort(arr, low, high):

    if low < high:

        pivot_index = randomized_partition(
            arr, low, high
        )

        randomized_quick_sort(
            arr, low, pivot_index - 1
        )

        randomized_quick_sort(
            arr, pivot_index + 1, high
        )


# ==========================================================
# 3. GENERATE TEST ARRAYS
# ==========================================================

N = 10000

random.seed(42)

configurations = {

    "Random Array":
        [random.randint(1, 100000)
         for _ in range(N)],

    "Already Sorted":
        list(range(N)),

    "Reverse Sorted":
        list(range(N, 0, -1)),

    "Many Duplicates":
        [random.randint(1, 10)
         for _ in range(N)]
}


# ==========================================================
# 4. RUN EXPERIMENT
# ==========================================================

print("QUICK SORT PERFORMANCE COMPARISON")
print("=================================")

print("\nArray Size:", N)

print("\nConfiguration Results")
print("--------------------")


for name, original_array in configurations.items():

    print("\n" + name)
    print("-" * len(name))


    # ------------------------------------------------------
    # Deterministic Quick Sort
    # ------------------------------------------------------

    arr1 = original_array.copy()

    det_comparisons = 0

    start_time = time.perf_counter()

    deterministic_quick_sort(
        arr1,
        0,
        len(arr1) - 1
    )

    det_time = time.perf_counter() - start_time


    # ------------------------------------------------------
    # Randomized Quick Sort
    # ------------------------------------------------------

    arr2 = original_array.copy()

    rand_comparisons = 0

    start_time = time.perf_counter()

    randomized_quick_sort(
        arr2,
        0,
        len(arr2) - 1
    )

    rand_time = time.perf_counter() - start_time


    # ------------------------------------------------------
    # Verify sorting
    # ------------------------------------------------------

    det_correct = arr1 == sorted(original_array)

    rand_correct = arr2 == sorted(original_array)


    # ------------------------------------------------------
    # Display
    # ------------------------------------------------------

    print(
        f"Deterministic Quick Sort:"
    )

    print(
        f"  Time        : {det_time:.6f} seconds"
    )

    print(
        f"  Comparisons : {det_comparisons}"
    )

    print(
        f"  Correct     : {det_correct}"
    )


    print(
        f"\nRandomized Quick Sort:"
    )

    print(
        f"  Time        : {rand_time:.6f} seconds"
    )

    print(
        f"  Comparisons : {rand_comparisons}"
    )

    print(
        f"  Correct     : {rand_correct}"
    )


# ==========================================================
# 5. IMPORTANT NOTE
# ==========================================================

print("\n\nCOMPLEXITY")
print("==========")

print("Deterministic Quick Sort:")
print("  Average Case : O(n log n)")
print("  Worst Case   : O(n^2)")

print("\nRandomized Quick Sort:")
print("  Expected Case: O(n log n)")
print("  Worst Case   : O(n^2)")
```

```python
# ==========================================================
# QUICK SORT PIVOT STRATEGY COMPARISON
#
# 1. Deterministic       -> Last element
# 2. Randomized          -> Random element
# 3. Median-of-Three     -> median(low, mid, high)
#
# Input size = 10,000
# Cases:
#   - Sorted
#   - Reverse Sorted
#   - Random
#
# Measures:
#   - Execution time
#   - Number of comparisons
# ==========================================================

import random
import time
import sys

# Prevent recursion-depth error for worst-case Quick Sort
sys.setrecursionlimit(30000)


# ==========================================================
# GLOBAL COMPARISON COUNTERS
# ==========================================================

det_comparisons = 0
rand_comparisons = 0
median_comparisons = 0


# ==========================================================
# 1. DETERMINISTIC QUICK SORT
#    Pivot = Last Element
# ==========================================================

def deterministic_partition(arr, low, high):

    global det_comparisons

    pivot = arr[high]

    i = low - 1

    for j in range(low, high):

        # Comparison: arr[j] <= pivot
        det_comparisons += 1

        if arr[j] <= pivot:

            i += 1

            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = (
        arr[high],
        arr[i + 1]
    )

    return i + 1


def deterministic_quick_sort(arr, low, high):

    if low < high:

        pivot_index = deterministic_partition(
            arr, low, high
        )

        deterministic_quick_sort(
            arr,
            low,
            pivot_index - 1
        )

        deterministic_quick_sort(
            arr,
            pivot_index + 1,
            high
        )


# ==========================================================
# 2. RANDOMIZED QUICK SORT
#    Pivot = Random Element
# ==========================================================

def randomized_partition(arr, low, high):

    global rand_comparisons

    # Select random pivot
    random_index = random.randint(low, high)

    # Move random pivot to the end
    arr[random_index], arr[high] = (
        arr[high],
        arr[random_index]
    )

    pivot = arr[high]

    i = low - 1

    for j in range(low, high):

        rand_comparisons += 1

        if arr[j] <= pivot:

            i += 1

            arr[i], arr[j] = (
                arr[j],
                arr[i]
            )

    arr[i + 1], arr[high] = (
        arr[high],
        arr[i + 1]
    )

    return i + 1


def randomized_quick_sort(arr, low, high):

    if low < high:

        pivot_index = randomized_partition(
            arr,
            low,
            high
        )

        randomized_quick_sort(
            arr,
            low,
            pivot_index - 1
        )

        randomized_quick_sort(
            arr,
            pivot_index + 1,
            high
        )


# ==========================================================
# 3. MEDIAN-OF-THREE
#
# Pivot = median of:
# arr[low], arr[mid], arr[high]
# ==========================================================

def median_of_three(arr, low, mid, high):

    global median_comparisons

    a = arr[low]
    b = arr[mid]
    c = arr[high]

    # Compare a and b
    median_comparisons += 1

    if a > b:

        a, b = b, a

    # Compare b and c
    median_comparisons += 1

    if b > c:

        b, c = c, b

    # Compare a and b
    median_comparisons += 1

    if a > b:

        a, b = b, a

    # b is the median
    return b


def median_partition(arr, low, high):

    global median_comparisons

    mid = (low + high) // 2

    # Find median value
    pivot_value = median_of_three(
        arr,
        low,
        mid,
        high
    )

    # Move median value to the end
    if arr[low] == pivot_value:

        pivot_index = low

    elif arr[mid] == pivot_value:

        pivot_index = mid

    else:

        pivot_index = high

    arr[pivot_index], arr[high] = (
        arr[high],
        arr[pivot_index]
    )

    pivot = arr[high]

    i = low - 1

    for j in range(low, high):

        median_comparisons += 1

        if arr[j] <= pivot:

            i += 1

            arr[i], arr[j] = (
                arr[j],
                arr[i]
            )

    arr[i + 1], arr[high] = (
        arr[high],
        arr[i + 1]
    )

    return i + 1


def median_three_quick_sort(arr, low, high):

    if low < high:

        pivot_index = median_partition(
            arr,
            low,
            high
        )

        median_three_quick_sort(
            arr,
            low,
            pivot_index - 1
        )

        median_three_quick_sort(
            arr,
            pivot_index + 1,
            high
        )


# ==========================================================
# 4. CREATE TEST ARRAYS
# ==========================================================

N = 10000

random.seed(42)

sorted_array = list(range(N))

reverse_array = list(range(N, 0, -1))

random_array = [
    random.randint(1, 100000)
    for _ in range(N)
]


test_cases = {
    "Sorted Array": sorted_array,
    "Reverse Sorted Array": reverse_array,
    "Random Array": random_array
}


# ==========================================================
# 5. RUN EXPERIMENT
# ==========================================================

print("QUICK SORT PIVOT STRATEGY COMPARISON")
print("====================================")

print("\nArray Size:", N)

print("\n")


for case_name, original_array in test_cases.items():

    print("=" * 60)
    print(case_name)
    print("=" * 60)


    # ------------------------------------------------------
    # DETERMINISTIC QUICK SORT
    # ------------------------------------------------------

    arr1 = original_array.copy()

    det_comparisons = 0

    start = time.perf_counter()

    deterministic_quick_sort(
        arr1,
        0,
        N - 1
    )

    det_time = time.perf_counter() - start


    # ------------------------------------------------------
    # RANDOMIZED QUICK SORT
    # ------------------------------------------------------

    arr2 = original_array.copy()

    rand_comparisons = 0

    random.seed(42)

    start = time.perf_counter()

    randomized_quick_sort(
        arr2,
        0,
        N - 1
    )

    rand_time = time.perf_counter() - start


    # ------------------------------------------------------
    # MEDIAN-OF-THREE QUICK SORT
    # ------------------------------------------------------

    arr3 = original_array.copy()

    median_comparisons = 0

    start = time.perf_counter()

    median_three_quick_sort(
        arr3,
        0,
        N - 1
    )

    median_time = time.perf_counter() - start


    # ------------------------------------------------------
    # VERIFY SORTING
    # ------------------------------------------------------

    correct1 = arr1 == sorted(original_array)

    correct2 = arr2 == sorted(original_array)

    correct3 = arr3 == sorted(original_array)


    # ------------------------------------------------------
    # DISPLAY RESULTS
    # ------------------------------------------------------

    print("\nDeterministic Quick Sort")
    print("------------------------")
    print(
        "Comparisons:",
        det_comparisons
    )
    print(
        "Time:",
        f"{det_time:.6f}",
        "seconds"
    )
    print(
        "Correct:",
        correct1
    )


    print("\nRandomized Quick Sort")
    print("---------------------")
    print(
        "Comparisons:",
        rand_comparisons
    )
    print(
        "Time:",
        f"{rand_time:.6f}",
        "seconds"
    )
    print(
        "Correct:",
        correct2
    )


    print("\nMedian-of-Three Quick Sort")
    print("-------------------------")
    print(
        "Comparisons:",
        median_comparisons
    )
    print(
        "Time:",
        f"{median_time:.6f}",
        "seconds"
    )
    print(
        "Correct:",
        correct3
    )


    # ------------------------------------------------------
    # Find algorithm with minimum comparisons
    # ------------------------------------------------------

    comparison_results = {
        "Deterministic": det_comparisons,
        "Randomized": rand_comparisons,
        "Median-of-Three": median_comparisons
    }

    best = min(
        comparison_results,
        key=comparison_results.get
    )

    print(
        "\nFewest comparisons:",
        best
    )

    print()


# ==========================================================
# 6. COMPLEXITY SUMMARY
# ==========================================================

print("\n")
print("=" * 60)
print("COMPLEXITY SUMMARY")
print("=" * 60)

print("""
Deterministic Quick Sort
  Pivot              : Last element
  Average Case       : O(n log n)
  Worst Case         : O(n^2)

Randomized Quick Sort
  Pivot              : Random element
  Expected Case      : O(n log n)
  Worst Case         : O(n^2)

Median-of-Three
  Pivot              : Median of low, mid, high
  Average Case       : O(n log n)
  Worst Case         : O(n^2)
""")
```

```python
# ==========================================================
# INTRO SORT vs STANDARD QUICK SORT
#
# IntroSort:
#   - Quick Sort normally
#   - If recursion depth > 2*log2(n), use Heap Sort
#
# Test:
#   1. Sorted array of 10,000 elements
#   2. Random array of 10,000 elements
#
# Measures:
#   - Execution time
#   - Number of comparisons
#   - Correctness
# ==========================================================

import random
import time
import math


# ==========================================================
# GLOBAL COMPARISON COUNTERS
# ==========================================================

quick_comparisons = 0
intro_comparisons = 0


# ==========================================================
# 1. STANDARD QUICK SORT
#    Pivot = Last Element
# ==========================================================

def quick_partition(arr, low, high):

    global quick_comparisons

    pivot = arr[high]

    i = low - 1

    for j in range(low, high):

        quick_comparisons += 1

        if arr[j] <= pivot:

            i += 1

            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = (
        arr[high],
        arr[i + 1]
    )

    return i + 1


def quick_sort(arr):

    # Stack contains sub-array ranges
    stack = [(0, len(arr) - 1)]

    while stack:

        low, high = stack.pop()

        if low >= high:
            continue

        pivot = quick_partition(
            arr,
            low,
            high
        )

        # Push larger and smaller parts
        # onto the stack.
        #
        # This avoids Python recursion-depth
        # problems on sorted arrays.

        if pivot - 1 - low > high - (pivot + 1):

            stack.append(
                (low, pivot - 1)
            )

            stack.append(
                (pivot + 1, high)
            )

        else:

            stack.append(
                (pivot + 1, high)
            )

            stack.append(
                (low, pivot - 1)
            )


# ==========================================================
# 2. HEAP SORT USED BY INTROSORT
# ==========================================================

def heapify(arr, start, heap_size, root):

    global intro_comparisons

    while True:

        largest = root

        left = 2 * (root - start) + 1 + start
        right = 2 * (root - start) + 2 + start

        # Compare root with left child
        if left < start + heap_size:

            intro_comparisons += 1

            if arr[left] > arr[largest]:
                largest = left

        # Compare largest with right child
        if right < start + heap_size:

            intro_comparisons += 1

            if arr[right] > arr[largest]:
                largest = right

        if largest == root:
            break

        arr[root], arr[largest] = (
            arr[largest],
            arr[root]
        )

        root = largest


def heap_sort_range(arr, low, high):

    # Number of elements
    size = high - low + 1

    # Build max heap
    for i in range(
        low + size // 2 - 1,
        low - 1,
        -1
    ):

        heapify(
            arr,
            low,
            size,
            i
        )

    # Extract elements
    for end in range(
        high,
        low,
        -1
    ):

        arr[low], arr[end] = (
            arr[end],
            arr[low]
        )

        heapify(
            arr,
            low,
            end - low,
            low
        )


# ==========================================================
# 3. INTROSORT PARTITION
# ==========================================================

def intro_partition(arr, low, high):

    global intro_comparisons

    pivot = arr[high]

    i = low - 1

    for j in range(low, high):

        intro_comparisons += 1

        if arr[j] <= pivot:

            i += 1

            arr[i], arr[j] = (
                arr[j],
                arr[i]
            )

    arr[i + 1], arr[high] = (
        arr[high],
        arr[i + 1]
    )

    return i + 1


# ==========================================================
# 4. INTROSORT
# ==========================================================

def introsort(arr):

    if len(arr) <= 1:
        return

    # Maximum allowed recursion depth
    max_depth = int(
        2 * math.log2(len(arr))
    )

    # Use an explicit stack:
    # (low, high, depth_limit)

    stack = [
        (0, len(arr) - 1, max_depth)
    ]

    while stack:

        low, high, depth_limit = stack.pop()

        if low >= high:
            continue

        # --------------------------------------------------
        # DEPTH LIMIT EXCEEDED
        # Switch to Heap Sort
        # --------------------------------------------------

        if depth_limit <= 0:

            heap_sort_range(
                arr,
                low,
                high
            )

            continue

        # --------------------------------------------------
        # Normal Quick Sort partition
        # --------------------------------------------------

        pivot = intro_partition(
            arr,
            low,
            high
        )

        new_depth = depth_limit - 1

        # Add sub-arrays
        if pivot - 1 > low:

            stack.append(
                (
                    low,
                    pivot - 1,
                    new_depth
                )
            )

        if pivot + 1 < high:

            stack.append(
                (
                    pivot + 1,
                    high,
                    new_depth
                )
            )


# ==========================================================
# 5. CREATE TEST ARRAYS
# ==========================================================

N = 10000

random.seed(42)

sorted_array = list(range(N))

random_array = [
    random.randint(1, 100000)
    for _ in range(N)
]


# ==========================================================
# 6. TEST FUNCTION
# ==========================================================

def run_test(name, original_array):

    global quick_comparisons
    global intro_comparisons

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    # ------------------------------------------------------
    # Standard Quick Sort
    # ------------------------------------------------------

    quick_array = original_array.copy()

    quick_comparisons = 0

    start = time.perf_counter()

    quick_sort(quick_array)

    quick_time = time.perf_counter() - start

    quick_correct = (
        quick_array == sorted(original_array)
    )


    # ------------------------------------------------------
    # IntroSort
    # ------------------------------------------------------

    intro_array = original_array.copy()

    intro_comparisons = 0

    start = time.perf_counter()

    introsort(intro_array)

    intro_time = time.perf_counter() - start

    intro_correct = (
        intro_array == sorted(original_array)
    )


    # ------------------------------------------------------
    # Display Results
    # ------------------------------------------------------

    print("\nStandard Quick Sort")
    print("-------------------")

    print(
        "Execution Time :",
        f"{quick_time:.6f}",
        "seconds"
    )

    print(
        "Comparisons    :",
        quick_comparisons
    )

    print(
        "Correct        :",
        quick_correct
    )


    print("\nIntroSort")
    print("---------")

    print(
        "Execution Time :",
        f"{intro_time:.6f}",
        "seconds"
    )

    print(
        "Comparisons    :",
        intro_comparisons
    )

    print(
        "Correct        :",
        intro_correct
    )


    # ------------------------------------------------------
    # Improvement
    # ------------------------------------------------------

    if quick_time > 0:

        time_improvement = (
            (quick_time - intro_time)
            / quick_time
        ) * 100

        print(
            "\nTime Improvement of IntroSort:",
            f"{time_improvement:.2f}%"
        )


# ==========================================================
# 7. RUN TESTS
# ==========================================================

print("INTROSORT vs STANDARD QUICK SORT")
print("=================================")

print("Array Size:", N)

print(
    "\nIntroSort depth limit:",
    int(2 * math.log2(N))
)


run_test(
    "TEST 1: SORTED ARRAY",
    sorted_array
)

run_test(
    "TEST 2: RANDOM ARRAY",
    random_array
)


# ==========================================================
# 8. COMPLEXITY ANALYSIS
# ==========================================================

print("\n\n" + "=" * 60)
print("COMPLEXITY ANALYSIS")
print("=" * 60)

print("""
STANDARD QUICK SORT
-------------------
Best/Average : O(n log n)
Worst        : O(n^2)


HEAP SORT
---------
Best/Average/Worst : O(n log n)


INTROSORT
---------
Quick Sort is used normally.

When recursion depth exceeds:
    2 * log2(n)

the algorithm switches to Heap Sort.

Therefore:

Worst Case : O(n log n)
Average    : O(n log n)
Best       : O(n log n) approximately
""")
```
