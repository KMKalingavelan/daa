```python
import heapq

V = 6

# Function to print the shortest path
def print_path(parent, vertex):
    path = []
    while vertex != -1:
        path.append(vertex)
        vertex = parent[vertex]
    print(" -> ".join(map(str, path[::-1])))

# Dijkstra's Algorithm
def dijkstra(graph, source):
    dist = [float('inf')] * V
    parent = [-1] * V

    dist[source] = 0
    pq = [(0, source)]  # (distance, vertex)

    while pq:
        current_dist, u = heapq.heappop(pq)

        if current_dist > dist[u]:
            continue

        for v, weight in graph[u]:
            if dist[u] + weight < dist[v]:
                dist[v] = dist[u] + weight
                parent[v] = u
                heapq.heappush(pq, (dist[v], v))

    # Print results
    print("Source Vertex:", source)
    print()

    for i in range(V):
        print(f"Destination: {i}")

        if dist[i] == float('inf'):
            print("Shortest Distance: INF")
            print("Path: No Path")
        else:
            print("Shortest Distance:", dist[i])
            print("Path:", end=" ")
            print_path(parent, i)

        print("-" * 30)


# Directed weighted graph
graph = [[] for _ in range(V)]

graph[0].append((1, 4))
graph[0].append((2, 2))

graph[1].append((2, 5))
graph[1].append((3, 10))

graph[2].append((4, 3))

graph[4].append((3, 4))
graph[4].append((5, 2))

graph[3].append((5, 11))

source = 0
dijkstra(graph, source)
```
