```python
import random
comparison_count = 0 # Global counter
def min_max_dc(arr, low, high):
    global comparison_count
    # Base case: single element
    if low == high:
        return arr[low], arr[low]
    # Base case: two elements
    if high == low + 1:
        comparison_count += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high]
        return arr[high], arr[low]
    # Divide
    mid = (low + high) // 2
    lmin, lmax = min_max_dc(arr, low, mid)
    rmin, rmax = min_max_dc(arr, mid + 1, high)
    # Conquer: combine with 2 comparisons
    comparison_count += 1
    overall_min = lmin if lmin < rmin else rmin
    comparison_count += 1
    overall_max = lmax if lmax > rmax else rmax
    return overall_min, overall_max
def min_max_naive(arr):
    mn, mx = arr[0], arr[0]
    comps = 0
    for x in arr[1:]:
        comps += 1
        if x < mn: mn = x
        comps += 1
        if x > mx: mx = x
    return mn, mx, comps
# --- Demonstration on small array ---
arr = [3, 1, 7, 4, 9, 2, 8, 5, 6, 0]
comparison_count = 0
mn, mx = min_max_dc(arr, 0, len(arr) - 1)
dc_comps = comparison_count
_, _, naive_comps = min_max_naive(arr)
print(f'Array: {arr}')
print(f'Min: {mn}, Max: {mx}')
print(f'D&C Comparisons: {dc_comps}')
print(f'Naive Comparisons: {naive_comps}')
# --- Performance Analysis ---
print(f'\n{"Size":>8} {"DC Comps":>12} {"Naive Comps":>14} {"Formula 3n/2-2":>16}')
print('-' * 56)
for size in [10, 100, 1000, 10000]:
    arr = [random.randint(1, 10000) for _ in range(size)]
    comparison_count = 0
    mn, mx = min_max_dc(arr, 0, len(arr) - 1)
    dc = comparison_count
    _, _, naive = min_max_naive(arr)
    formula = 3 * size // 2 - 2
    print(f'{size:>8} {dc:>12} {naive:>14} {formula:>16}')
```

```python
# Divide and Conquer Min-Max with Second Minimum and Second Maximum

def find_min_max_second(arr, low, high):

    # Base case: only one element
    if low == high:
        return arr[low], float('inf'), arr[low], float('-inf')

    # Base case: two elements
    if high == low + 1:

        if arr[low] < arr[high]:
            return arr[low], arr[high], arr[high], arr[low]
        else:
            return arr[high], arr[low], arr[low], arr[high]

    # Find middle
    mid = (low + high) // 2

    # Divide
    min1, second_min1, max1, second_max1 = find_min_max_second(
        arr, low, mid
    )

    min2, second_min2, max2, second_max2 = find_min_max_second(
        arr, mid + 1, high
    )

    # ------------------------------------------------
    # Combine
    # ------------------------------------------------

    # Find minimum and second minimum
    if min1 < min2:
        minimum = min1
        second_minimum = min(second_min1, min2)
    else:
        minimum = min2
        second_minimum = min(min1, second_min2)

    # Find maximum and second maximum
    if max1 > max2:
        maximum = max1
        second_maximum = max(second_max1, max2)
    else:
        maximum = max2
        second_maximum = max(max1, second_max2)

    return minimum, second_minimum, maximum, second_maximum


# ------------------------------------------------
# Test the algorithm
# ------------------------------------------------

arr = [12, 5, 8, 20, 3, 15, 7, 25]

n = len(arr)

minimum, second_minimum, maximum, second_maximum = \
    find_min_max_second(arr, 0, n - 1)


# ------------------------------------------------
# Display Result
# ------------------------------------------------

print("Divide and Conquer Min-Max Algorithm")
print("--------------------------------------")

print("Array:", arr)

print("\nMinimum Element       :", minimum)
print("Second Minimum        :", second_minimum)
print("Maximum Element       :", maximum)
print("Second Maximum        :", second_maximum)
```

```python
# QuickSelect - Divide and Conquer Algorithm
# Finding the k-th smallest element in an unsorted array

import random


# --------------------------------------------------
# Partition function
# --------------------------------------------------

def partition(arr, low, high):

    # Choose the last element as pivot
    pivot = arr[high]

    i = low

    # Place elements smaller than pivot on the left
    for j in range(low, high):

        if arr[j] <= pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1

    # Place pivot in its correct position
    arr[i], arr[high] = arr[high], arr[i]

    return i


# --------------------------------------------------
# QuickSelect function
# --------------------------------------------------

def quickselect(arr, low, high, k):

    # k is given as 1-based position
    # Convert it to 0-based index
    k_index = k - 1

    while low <= high:

        # Partition the array
        pivot_index = partition(arr, low, high)

        # Pivot is exactly the k-th smallest element
        if pivot_index == k_index:
            return arr[pivot_index]

        # Search in the left half
        elif k_index < pivot_index:
            high = pivot_index - 1

        # Search in the right half
        else:
            low = pivot_index + 1

    return None


# --------------------------------------------------
# Create an array of 1000 random integers
# --------------------------------------------------

n = 1000

arr = [random.randint(1, 10000) for _ in range(n)]

print("QuickSelect - Divide and Conquer")
print("---------------------------------")
print("Number of elements:", n)


# --------------------------------------------------
# Test 1: k = 1 (Minimum)
# --------------------------------------------------

arr1 = arr.copy()

k1 = 1

result_min = quickselect(
    arr1, 0, n - 1, k1
)

print("\nTest 1: k = 1")
print("1st Smallest Element (Minimum):", result_min)


# --------------------------------------------------
# Test 2: k = n (Maximum)
# --------------------------------------------------

arr2 = arr.copy()

kn = n

result_max = quickselect(
    arr2, 0, n - 1, kn
)

print("\nTest 2: k = n")
print("1000th Smallest Element (Maximum):", result_max)


# --------------------------------------------------
# Test 3: k = n//2 (Median)
# --------------------------------------------------

arr3 = arr.copy()

k_median = n // 2

result_median = quickselect(
    arr3, 0, n - 1, k_median
)

print("\nTest 3: k = n//2")
print("500th Smallest Element (Median):", result_median)


# --------------------------------------------------
# Verification using Python's sorted()
# --------------------------------------------------

sorted_arr = sorted(arr)

print("\nVerification")
print("------------")
print("Minimum :", sorted_arr[0])
print("Maximum :", sorted_arr[-1])
print("500th smallest :", sorted_arr[499])
```
